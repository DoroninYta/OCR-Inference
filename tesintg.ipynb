```python
cap = cv.VideoCapture(folder / "7.avi")
if cap.isOpened() == False:
    print("Error in opening video stream or file")
while(cap.isOpened()):
    ret, frame = cap.read()
    if ret:
        frame = cv.cvtColor(frame, cv.COLOR_BGR2RGB)
        result = model.track(frame, stream=False, classes=2)
        for res in result:
            print(res)
        plt.imshow(result[0].plot())
        plt.show()
        # Press esc to exit
        if cv.waitKey(20) & 0xFF == 27:
            break
    else:
        break
cap.release()
cv.destroyAllWindows()
```
